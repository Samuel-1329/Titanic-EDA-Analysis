Observation:

1. The dataset contains 891 records and 12 features.
2. Missing values are present in Age, Cabin, and Embarked columns.
3. Cabin contains the highest number of missing values.
4. Numerical columns include Age, Fare, SibSp, Parch, etc.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')

plt.show()
```

Observation:

1. Most passengers were between 20 and 40 years old.
2. The age distribution is slightly right-skewed.
3. There were fewer elderly passengers compared to adults.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df)

plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')

plt.show()
```

Observation:

1. More passengers died than survived.
2. The survival rate was lower than the death rate.
3. The Titanic disaster resulted in significant loss of life.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='Sex', hue='Survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

Observation:

1. Female passengers had a much higher survival rate.
2. Male passengers experienced a higher mortality rate.
3. Gender appears to be an important factor affecting survival.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')

plt.show()
```

Observation:

1. First-class passengers had the highest survival rate.
2. Third-class passengers experienced the highest mortality rate.
3. Passenger class had a significant impact on survival chances.

```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include='number')

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

Observation:

1. Survival has a positive correlation with Fare.
2. Survival has a negative correlation with Pclass.
3. Fare and Pclass are negatively correlated.
4. Age shows a weak correlation with Survival.

```python
plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include='number')

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

Observation:

1. Survival has a positive correlation with Fare.
2. Survival has a negative correlation with Passenger Class (Pclass).
3. Fare and Pclass are negatively correlated.
4. Age has a weak correlation with Survival.
5. No strong multicollinearity is observed among numerical features.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Survived',
    y='Fare',
    data=df
)

plt.title('Fare vs Survival')
plt.xlabel('Survived')
plt.ylabel('Fare')

plt.show()
```

Observation:

1. Survivors generally paid higher fares.
2. Higher ticket fare appears to be associated with better survival chances.
3. Several outliers exist in the Fare column.

```python
sns.pairplot(
    df[['Survived','Age','Fare','Pclass']]
)

plt.show()
```

Observation:

1. Fare and Passenger Class show a noticeable relationship.
2. Survival is more closely related to Fare and Pclass than Age.
3. Age does not show a strong separation between survivors and non-survivors.
